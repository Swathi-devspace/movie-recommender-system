```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv("tmdb_5000_movies.csv")
credits=pd.read_csv("tmdb_5000_credits.csv")
```

```python
movies=movies.merge(credits,on='title')
```

```python
#genres
#id
#keywords
#title
#overview
#cast
#crew
movies=movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.isnull().sum
```

```python
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
import ast
```

```python
def convert(text):
    my_list=[]
    for i in ast.literal_eval(text):
        my_list.append(i['name'])
    return my_list
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
def convert1(text):
    my_list=[]
    count=0
    for i in ast.literal_eval(text):
        if count>=3:
            break
        my_list.append(i['name'])
        count+=1
    return my_list
```

```python
movies['cast']=movies['cast'].apply(convert1)
```

```python
def retrieve_director(text):
    my_list=[]
    for i in ast.literal_eval(text):
        if i['job']=='Director':
            my_list.append(i['name'])
            break
    return my_list
```

```python
movies['crew']=movies['crew'].apply(retrieve_director)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['tags']=movies['overview']+movies['keywords']+movies['cast']+movies['crew']+movies['genres']
```

```python
new_movies_df=movies[['movie_id','title','tags']]
```

```python
new_movies_df
```

```python
new_movies_df['tags']=new_movies_df['tags'].apply(lambda x:' '.join(x))
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
pt=PorterStemmer()
```

```python
def stem_text(text):
    L=[]
    for i in text.split():
        L.append(pt.stem(i))
    return ' '.join(L)
```

```python
new_movies_df['tags']=new_movies_df['tags'].apply(stem_text)
```

```python
new_movies_df
```

```python
new_movies_df['tags']=new_movies_df['tags'].apply(lambda x: x.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_movies_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
similarity = similarity.astype('float32')
```

```python
def recommend(movie):
    movie_index=new_movies_df[new_movies_df['title']==movie].index[0]
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_movies_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
